```python
spark.sql("CREATE SCHEMA IF NOT EXISTS ibm_hr.silver")

spark.sql("""
CREATE OR REPLACE TABLE ibm_hr.silver.dim_employee AS
SELECT
    EmployeeNumber AS employee_id,
    Age,
    Gender,
    MaritalStatus,
    Department,
    JobRole,
    JobLevel,
    Education,
    EducationField,
    BusinessTravel,
    OverTime,
    Attrition
FROM ibm_hr.bronze.employees_raw
""")

spark.sql("""
CREATE OR REPLACE TABLE ibm_hr.silver.fact_employee_metrics AS
SELECT
    EmployeeNumber AS employee_id,
    MonthlyIncome,
    DailyRate,
    HourlyRate,
    MonthlyRate,
    JobSatisfaction,
    EnvironmentSatisfaction,
    WorkLifeBalance,
    RelationshipSatisfaction,
    PerformanceRating,
    PercentSalaryHike,
    YearsAtCompany,
    YearsInCurrentRole,
    YearsSinceLastPromotion,
    YearsWithCurrManager,
    TotalWorkingYears,
    NumCompaniesWorked,
    TrainingTimesLastYear,
    StockOptionLevel
FROM ibm_hr.bronze.employees_raw
""")
```

```python
spark.sql("SELECT * FROM ibm_hr.silver.dim_employee LIMIT 5").show()
spark.sql("SELECT * FROM ibm_hr.silver.fact_employee_metrics LIMIT 5").show()
```
