```python
#reading the CSV into a Spark DataFrame and taking a look at it.
df = spark.read.option("header", "true").option("inferSchema", "true").csv(
    "/Volumes/ibm_hr/source/source_volume/WA_Fn-UseC_-HR-Employee-Attrition.csv"
)

display(df)
print(f"Row count: {df.count()}")
df.printSchema()
```
